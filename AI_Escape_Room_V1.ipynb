```python

GAME_PROMPT = """
You are the Game Master of an interactive escape room
called THE MYSTERY OF THE LOCKED LIBRARY.

YOUR ROLE:
Guide the player through an exciting escape room adventure.
Respond naturally to the player's messages.
Keep track of the conversation and previous discoveries.

ROOM ENVIRONMENT:
The player is trapped inside a mysterious library.

There are exactly five interactive objects:
1. Bookshelf
2. Desk
3. Painting
4. Locked drawer
5. Exit door

PUZZLE DESIGN:

PUZZLE 1 - BOOKSHELF:
Four unusual books contain numbers.
When properly investigated, they reveal the drawer code: 2413.

PUZZLE 2 - PAINTING:
A riddle guides the player to investigate behind the painting.
The player discovers a brass key.

PUZZLE 3 - LOCKED DRAWER:
The drawer requires the brass key and the correct code 2413.
Inside the drawer is a note containing the final door code: 7391.

FINAL CHALLENGE:
The exit door requires the four-digit code 7391.

IMPORTANT RULES:
- Do not reveal puzzle answers immediately.
- Give hints when the player asks for help.
- Do not invent additional interactive objects.
- Remember what the player has discovered.
- Do not claim that an item was collected unless the
  player actually completed the relevant action.
- Give incorrect-code attempts an appropriate response.
- Keep the story engaging and consistent.
- Allow the player to restart by requesting a new game.
- The player must complete the puzzles before escaping.

VERSION 1 LIMITATION:
This version uses AI conversation to manage the game.
It does not have strict Python-based action validation.
The AI may occasionally forget or misunderstand actions.

START:
Welcome the player to the mysterious library.
Describe the five objects and explain how to play.
Do not reveal the secret codes.
"""

print("Game instructions created successfully!")
```

```python

MODEL_NAME = "gemini-3.8-flash"

def game_master(message, history):

    conversation = GAME_PROMPT + "\n\n"

    # Include previous conversation
    for item in history:
        role = item.get("role", "user")
        content = item.get("content", "")

        if isinstance(content, str):
            conversation += f"{role.upper()}: {content}\n"

    # Add current player message
    conversation += f"PLAYER: {message}\n"
    conversation += "GAME MASTER:"

    try:
        response = client.models.generate_content(
            model=MODEL_NAME,
            contents=conversation
        )

        return response.text

    except Exception as e:
        return f"An error occurred: {str(e)}"

print("Game Master function created successfully!")
```

```python

import gradio as gr

demo = gr.ChatInterface(
    fn=game_master,
    type="messages",
    title="THE MYSTERY OF THE LOCKED LIBRARY",
    description="""
    You are trapped inside a mysterious library.

    Explore five objects, discover hidden clues,
    solve three puzzles and unlock the exit door.

    Talk naturally to the AI Game Master.
    Good luck!
    """,
    examples=[
        "I want to examine the bookshelf.",
        "Can I investigate the painting?",
        "I want to inspect the desk.",
        "How can I escape this library?"
    ],
    textbox=gr.Textbox(
        placeholder="What would you like to do?",
        container=True
    )
)

print("Escape Room interface created successfully!")
```

```python

MODEL_NAME = "gemini-3.5-flash-lite"

import time

def game_master(message, history):

    conversation = GAME_PROMPT + "\n\n"

    for item in history:
        role = item.get("role", "user")
        content = item.get("content", "")

        if isinstance(content, str):
            conversation += f"{role.upper()}: {content}\n"

    conversation += f"PLAYER: {message}\n"
    conversation += "GAME MASTER:"

    for attempt in range(3):
        try:
            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=conversation
            )
            return response.text

        except Exception as e:
            if "503" in str(e) or "UNAVAILABLE" in str(e):
                if attempt < 2:
                    time.sleep(5 * (attempt + 1))
                else:
                    return "The library is temporarily unavailable. Please try again."
            else:
                return f"Error: {str(e)}"

print("Game Master updated successfully!")
```

```python

import gradio as gr

demo = gr.ChatInterface(
    fn=game_master,
    title="THE MYSTERY OF THE LOCKED LIBRARY",
    description="""
    Welcome to the mysterious library!

    Explore five objects, discover hidden clues,
    solve three puzzles and unlock the exit door.

    Talk naturally to your AI Game Master.
    """
)

demo.launch(share=True, debug=True)
```
